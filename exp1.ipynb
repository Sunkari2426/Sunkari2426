```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler, StandardScaler, LabelEncoder, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer

data = {
    'Age':[25, 30, np.nan, 45, 22],
    'Salary':[50000, 60000, 52000, np.nan, 58000],
    'Department':['HR', 'Finance', 'HR', 'IT', np.nan],
    'Gender':['Male','Female','Female', np.nan, 'Male']
}

df = pd.DataFrame(data)
print("Original Data")
print(df)


#Handling the Missing values
#Numerical columns fill missing values with mean

num_imputer = SimpleImputer(strategy='mean')
df[['Age', 'Salary']] = num_imputer.fit_transform(df[['Age', 'Salary']])


#Categorical columns fill missing values with mode
cat_imputer = SimpleImputer(strategy='most_frequent')
df[['Department', 'Gender']] = cat_imputer.fit_transform(df[['Department', 'Gender']])
print("\n\nData after handling missing values\n")
print(df)


#Normalization (Min-Max Scalling)
scaler = MinMaxScaler()
df[['Age', 'Salary']] = scaler.fit_transform(df[['Age', 'Salary']])
print("\n\nData after normalization\n")
print(df)

#Standerdization (z-score scaling)
scaler = StandardScaler()
df[['Age', 'Salary']] = scaler.fit_transform(df[['Age', 'Salary']])
print("\n\nData after standerdization\n")
print(df)

#Encoding Categorical variables
#Label Encoding for 'Gender'
le = LabelEncoder()
df['Gender'] = le.fit_transform(df['Gender'])

#One-Hot Encoding for 'Department'
df = pd.get_dummies(df, columns=['Department'])
print("\n\nData after encoding\n")
print(df)

```
